```python
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedGroupKFold
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

veri_yolu = "/content/iris.csv"
data = pd.read_csv(veri_yolu)
data.head()
```

```python
data.drop('Id',axis=1,inplace =True)
```

```python
from sklearn.preprocessing import LabelEncoder as labelEncoder
features = ['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
x = data.drop('Species',axis=1).values
y = data['Species'].values
le = labelEncoder()
y = le.fit_transform(y)
print("Sınıf Labelleri: ",list(le.classes_))
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.4,random_state=42) #start
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
x_train = pd.DataFrame(x_train,columns=features)
x_test = pd.DataFrame(x_test,columns=features)
```

```python
x_train_dirty = x_train.copy()
missing_rows = x_train_dirty.index[0:2]# îlk iki satır
outlier_rows = x_train_dirty.index[2:4] #son iki satır

affected_rows =list(missing_rows) + list(outlier_rows)
```

```python
print("orijinal veriler:")
print(x_train.loc[affected_rows])

x_train_dirty.loc[missing_rows,'SepalLengthCm'] = np.nan
x_train_dirty.loc[missing_rows,'PetalLengthCm'] = np.nan
x_train_dirty.loc[outlier_rows,'SepalWidthCm'] = 100.0
x_train_dirty.loc[outlier_rows,'PetalWidthCm'] = 200.0 #bu şekilde iki aykırı veri 2 boş veri oldu
```

```python
print('Sonra:')
print(x_train_dirty.loc[affected_rows])
```

```python
x_train_dirty.isnull().sum()
```

```python
q1 = x_train_dirty.quantile(0.25)
q3 = x_train_dirty.quantile(0.75)
IQR = q3 - q1
lower = q1 - 1.5 * IQR
upper = q3 + 1.5 * IQR #upper ve lowe arasında ise iyi değilse aykırı deriz
outlier_mask = x_train_dirty.lt(lower) | x_train_dirty.gt(upper)
print("----")
print(x_train_dirty[outlier_mask.any(axis=1)]) #aykırıları getirir
```

```python
x_train_clean = x_train_dirty.clip(lower,upper,axis=1) #clip mean ile ortalama değerini alıp atamasını yapıyor
fill_values = x_train_clean.mean()
x_train_clean = x_train_clean.fillna(fill_values)
print("Temizlenmiş Veriler:")
print(x_train_clean.loc[affected_rows])
```

```python
outlier_mask = x_train_clean.lt(lower) | x_train_clean.gt(upper)
print("Temizlenmiş Veriler:")
print(x_train_clean[outlier_mask.any(axis=1)])
```

```python
x_train_clean.isnull().sum()
```

```python
from sklearn.metrics import classification_report
knn_clean = KNeighborsClassifier(n_neighbors=3)
knn_clean.fit(x_train_clean,y_train)
y_pred_clean = knn_clean.predict(x_test)
print(pd.DataFrame({'Gerçek':y_test,'Tahmin':y_pred_clean}))
print("Doğruluk Oranı: {:.2f}".format(accuracy_score(y_test,y_pred_clean)))
print("Sınıflandırma Raporu:",classification_report(y_test,y_pred_clean))
```
