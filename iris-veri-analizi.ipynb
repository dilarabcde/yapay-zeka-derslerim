```python
df_yeni = df.drop(columns=["Id"])
df_yeni.head(3)
```

```python
import matplotlib.pyplot as plt

features = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']

df_yeni[features].plot(figsize=(10, 4))
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

setosa = df[df['Species'] == 'Iris-setosa']

plt.figure(figsize=(10, 4))
plt.plot(setosa['Id'], setosa['SepalLengthCm'], label='setosa')
plt.xlabel('Id')
plt.ylabel('SepalLengthCm')
plt.legend()
plt.show()
```

```python
groups = [('setosa', 'red', 'setosa'),
          ('versicolor', 'blue', 'versicolor'),
          ('virginica', 'green', 'virginica')]

for feature in ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']:
    plt.figure(figsize=(10, 4))

    plt.plot(setosa['Id'], setosa[feature], color='red', label='setosa')
    plt.plot(versicolor['Id'], versicolor[feature], color='blue', label='versicolor')
    plt.plot(virginica['Id'], virginica[feature], color='green', label='virginica')

    plt.xlabel('Id')
    plt.ylabel(f'{feature} (cm)')
    plt.title(f'{feature} by species')
    plt.legend()
    plt.show()
```

```python
for x_feature in ['SepalLengthCm', 'PetalLengthCm']:
    plt.figure(figsize=(10, 4))
    for subset, color, label in groups:
        subset_df = df[df['Species'] == 'Iris-' + subset]
        plt.scatter(subset_df[x_feature], subset_df['PetalWidthCm'], color=color, label=label)
    plt.xlabel(x_feature)
    plt.ylabel('PetalWidthCm')
    plt.title(f'{x_feature} vs. PetalWidthCm')
    plt.legend()
    plt.show()
```
